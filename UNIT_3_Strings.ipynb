```python
name = "Arun"
city = 'Coimbatore'
message = """Welcome to Python Programming"""

print(name)
print(city)
print(message)
```

```python
#String Operations in Python
#String concatenation means joining two or more strings using the + operator.
first_name = "Arun"
last_name = "Kumar"
full_name = first_name + " " + last_name
print(full_name)
```

```python
#String Repetition:String repetition means repeating a string multiple times using the * operator.
name = "Python "
print(name * 3)
```

```python
#5. String Indexing

#Indexing means accessing individual characters from a string using their position.

#Python supports two types of indexing:

#    Positive indexing

#    Negative indexing

#A. Positive Indexing

#Positive indexing starts from 0 (left to right).
name = "PYTHON"

print(name[0])
print(name[1])
print(name[2])
print(name[3])
```

```python
#B. Negative Indexing:Negative indexing starts from -1 (right to left).
name = "PYTHON"
print(name[-1])
print(name[-2])
print(name[-3])
```

```python
#6. String Slicing:String slicing means extracting a part of a string.
#Syntax:
#string[start:end:step]
name = "PYTHON PROGRAMMING"

print(name[0:6])
print(name[7:18])
print(name[:6])
print(name[7:])
print(name[::-1])
```

Explanation:

    name[0:6] – extracts characters from index 0 to 5.

    name[7:18] – extracts characters from index 7 to 17.

    name[:6] – extracts from beginning to index 5.

    name[7:] – extracts from index 7 to the end.

    name[::-1] – reverses the string.

```python
#String Length:The len() function is used to find the total number of characters in a string, including spaces.
name = "Python"
print(len(name))
```

```python
#String Membership Operators:Membership operators are used to check whether a character or substring exists in a string.
name = "Python Programming"

print("Python" in name)
print("Java" in name)
print("Java" not in name)
```

```python
#String Comparison:String comparison is used to compare two strings using comparison operators.
a = "Python"
b = "Python"
c = "Java"

print(a == b)
print(a != c)
print(a == c)
```

```python
#Important String Methods in Python
#A. upper():Converts all characters into uppercase.
name = "python"
print(name.upper())
```

```python
#B. lower():Converts all characters into lowercase.
name = "PYTHON"
print(name.lower())
```

```python
#C. title():Converts the first character of each word into uppercase.
name = "python programming"
print(name.title())
```

```python
#D. capitalize():Converts the first character of the string into uppercase.
name = "python programming"
print(name.capitalize())
```

```python
#E. strip():Removes spaces from the beginning and end of a string.
name = "  Python  "
print(name.strip())
```

```python
#F. replace():Replaces one substring with another.
text = "I like Java"
print(text.replace("Java", "Python"))
```

```python
#find():Returns the index of the first occurrence of a substring. Returns -1 if it is not found.
text = "Python Programming"
print(text.find("Program"))
```

```python
#H. count():Counts how many times a character or substring appears.
text = "banana"
print(text.count("a"))
```

```python
#I. split():Splits a string into a list.
text = "Python is easy"
print(text.split())
```

```python
#J. join():Joins multiple strings using a separator.
words = ["Python", "is", "easy"]
print(" ".join(words))
```

```python
#K. startswith():Checks whether a string starts with a particular substring.
text = "Python Programming"
print(text.startswith("Python"))
```

```python
#L. endswith():Checks whether a string ends with a particular substring.
text = "Python Programming"
print(text.endswith("Programming"))
```

```python
#M. isdigit():Checks whether all characters in a string are digits.
number = "12345"
print(number.isdigit())
```

```python
#N. isalpha():Checks whether all characters are alphabets.
name = "Python"
print(name.isalpha())
```

```python
#O. isalnum():Checks whether all characters are alphabets or numbers, with at least one character.
text = "Python123"
print(text.isalnum())
```

```python
#String Immutability:Strings in Python are immutable, which means we cannot directly change individual characters after creating a string.
name = "Python"

name[0] = "J"

print(name)
```

```python
#B. Import specific function
#We can import only the function that we need.
from math import sqrt
print(sqrt(36))
```
