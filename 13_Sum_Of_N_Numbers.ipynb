<font color="blue" size="6"> Sum of first N natural numbers</font>

```python
n = int(input("Enter a number: "))

sum = 0

for i in range(1, n + 1):
    sum += i

print("Sum =", sum)
```

```python
#Shortcut using the formula:
n = int(input("Enter a number: "))
sum = n * (n + 1) // 2
print("Sum =", sum)
```
