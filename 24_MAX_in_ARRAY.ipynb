<font color="blue" size="6">Find maximum and minimum in an array</font>

```python
# Get the number of elements
n = int(input("Enter the number of elements: "))

# Create an array
arr = []

# Get the values
for i in range(n):
    value = int(input(f"Enter element {i + 1}: "))
    arr.append(value)

# Find maximum and minimum
maximum = max(arr)
minimum = min(arr)

# Print the results
print("Array:", arr)
print("Maximum:", maximum)
print("Minimum:", minimum)
```

```python
#Here is how to find the maximum and minimum without using max() or min().
# Get the number of elements
n = int(input("Enter the number of elements: "))

# Create an array
arr = []

# Get the values
for i in range(n):
    value = int(input(f"Enter element {i + 1}: "))
    arr.append(value)

# Assume the first element is both maximum and minimum
maximum = arr[0]
minimum = arr[0]

# Compare the remaining elements
for i in range(1, n):
    if arr[i] > maximum:
        maximum = arr[i]

    if arr[i] < minimum:
        minimum = arr[i]

print("Array:", arr)
print("Maximum:", maximum)
print("Minimum:", minimum)
```
