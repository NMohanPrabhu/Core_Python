<font color="blue" size="6">merge two arrays without using a built-in function</font>

```python
# Get first array
n1 = int(input("Enter number of elements in first array: "))
arr1 = []

for i in range(n1):
    value = int(input(f"Enter element {i + 1}: "))
    arr1.append(value)

# Get second array
n2 = int(input("Enter number of elements in second array: "))
arr2 = []

for i in range(n2):
    value = int(input(f"Enter element {i + 1}: "))
    arr2.append(value)

# Create merged array
merged = []

# Copy first array
for i in range(n1):
    merged.append(arr1[i])

# Copy second array
for i in range(n2):
    merged.append(arr2[i])

print("First array:", arr1)
print("Second array:", arr2)
print("Merged array:", merged)
```
