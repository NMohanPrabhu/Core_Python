```python
#Student Marks Statement Using List and Tuple in Python
#A student marks statement is a program used to store student details and subject marks, calculate the total and average marks, and display the result.
```

```python
# Student details using tuple
student = ("Ravi", 101)

# Subject marks using list
marks = [85, 90, 75, 80, 95]

# Calculate total and average
total = sum(marks)
average = total / len(marks)

# Display student marks statement
print("STUDENT MARKS STATEMENT")
print("-----------------------")

print("Student Name:", student[0])
print("Register Number:", student[1])

print("Marks:", marks)
print("Total Marks:", total)
print("Average Marks:", average)

if average >= 50:
    print("Result: PASS")
else:
    print("Result: FAIL")
```

Explanation

    student = ("Ravi", 101) – Creates a tuple containing student details.

    marks = [85, 90, 75, 80, 95] – Creates a list containing five subject marks.

    sum(marks) – Calculates the total marks.

    len(marks) – Counts the number of subjects.

    average = total / len(marks) – Calculates the average.

    if average >= 50 – Checks whether the student passes.
