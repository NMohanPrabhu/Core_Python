<font color="blue" size="6">calculate the sum and average by traversing the array</font>

```python
# Get the number of elements
n = int(input("Enter the number of elements: "))

# Create an array
arr = []

# Get the values
for i in range(n):
    value = int(input(f"Enter element {i + 1}: "))
    arr.append(value)

# Calculate sum
total = 0

for i in range(n):
    total = total + arr[i]

# Calculate average
average = total / n

print("Array:", arr)
print("Sum:", total)
print("Average:", average)
```
