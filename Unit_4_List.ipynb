<b>AIM:TO study various operations that can be performed on a list. 

<b>Python lists are ordered, mutable (changeable), and allow duplicate elements.
<p>In Python, list operations are used to create, access, add, remove, modify, and manipulate elements in a list.


```python
#1. Creating a list
fruits = ["apple", "banana", "orange"]
print(fruits)
```

2. Basic list operations


```python
#3. Accessing list elements
numbers = [10, 20, 30, 40, 50]

print(numbers[0])   # 10
print(numbers[2])   # 30
print(numbers[-1])  # 50
```

```python
#4. List slicing
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])  # [20, 30, 40]
print(numbers[:3])   # [10, 20, 30]
print(numbers[2:])   # [30, 40, 50]
print(numbers[::-1]) # Reverses the list
```

```python
#5. Updating list elements
numbers = [10, 20, 30]
numbers[1] = 100
print(numbers)
# [10, 100, 30]
```

```python
#6. Mathematical operations on lists
a = [1, 2, 3]
b = [4, 5, 6]

print(a + b)  # Concatenation
print(a * 2)  # Repetition
print(len(a)) # Length
print(max(a)) # Maximum
print(min(a)) # Minimum
print(sum(a)) # Sum
```

```python
#7. Checking elements in a list
numbers = [10, 20, 30, 40]

print(20 in numbers)      # True
print(50 not in numbers)  # True
```

```python
#8. Iterating through a list
numbers = [10, 20, 30, 40]

for num in numbers:
    print(num)
```

```python
#9. List comprehension
numbers = [1, 2, 3, 4, 5]

squares = [x**2 for x in numbers]

print(squares)
# [1, 4, 9, 16, 25]
```

```python
#Accessing Individual Elements of a List by Indexing in Python
#$Indexing is used to access individual elements in a list using their position (index number).
#In Python, indexing starts from 0.
#1. Positive Indexing
#Positive indexing starts from left to right, beginning with index 0.
fruits = ["Apple", "Banana", "Mango", "Orange"]

print(fruits[0])  # Apple
print(fruits[1])  # Banana
print(fruits[2])  # Mango
print(fruits[3])  # Orange
```

```python
#2. Negative Indexing

#Negative indexing starts from right to left. The last element has index -1.
fruits = ["Apple", "Banana", "Mango", "Orange"]

print(fruits[-1])  # Orange
print(fruits[-2])  # Mango
print(fruits[-3])  # Banana
print(fruits[-4])  # Apple
```

```python
#3. Updating Elements Using Indexing

#We can also change a list element using its index.
numbers = [10, 20, 30, 40]

numbers[1] = 100

print(numbers)
```

Important Points

    List indexing starts from 0.

    Positive indexing moves from left to right.

    Negative indexing moves from right to left.

    The last element can be accessed using -1.

    Accessing an invalid index produces an IndexError.

    Indexing can be used to access and modify list elements.

<B>Getting Parts of a List by Slicing in Python

<P>#List slicing is used to access a specific portion or group of elements from a list.
<P>#It allows us to retrieve multiple elements without accessing them individually.
<P>#1. Syntax of List Slicing
<P>list_name[start : stop : step]
<P>start: Starting index (included).

<P>stop: Ending index (excluded).

<P>step: Number of positions to skip (optional).

```python
#2. Example of List Slicing
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]

print(fruits[1:4])
```

```python
#3. Different Types of Slicing
#a) Slicing from the beginning
numbers = [10, 20, 30, 40, 50]

print(numbers[:3])
```

```python
#b) Slicing up to the end
print(numbers[2:])
```

```python
#c) Slicing the entire list
print(numbers[:])
```

```python
#d) Slicing with step
print(numbers[::2])
```

```python
#e) Reversing a list using slicing
print(numbers[::-1])
```

Important Points

    Slicing returns multiple elements from a list.

    The starting index is included.

    The stopping index is excluded.

    Step determines the interval between elements.

    Negative indexes can also be used in slicing.

    Slicing does not modify the original list; it creates a new list.

<B>Changing an Item in a List by Index Assignment in Python

<P>Index assignment is used to change or modify an existing element in a list by specifying its index position.

<P>Python lists are mutable, which means their elements can be changed after creation.

<FONT SIZE="4" COLOR="RED"><B>Syntax:list_name[index] = new_value</FONT>
<P>list_name: Name of the list.
<P>index: Position of the element to be changed.
<P>new_value: New value assigned to that position.

```python
fruits = ["Apple", "Banana", "Mango", "Orange"]

fruits[1] = "Grapes"

print(fruits)
```

Explanation:

    Initially, the element at index 1 is "Banana".

    Using fruits[1] = "Grapes", we replace "Banana" with "Grapes".

    The other elements remain unchanged.

```python
#3. Changing Multiple Items
numbers = [10, 20, 30, 40, 50]

numbers[0] = 100
numbers[2] = 300
numbers[4] = 500

print(numbers)
```

```python
#4. Changing an Item Using Negative Indexing
colors = ["Red", "Green", "Blue"]

colors[-1] = "Yellow"

print(colors)
```

<font color="red" size="6">Shallow copy and deep copy</font>

<p>In Python, shallow copy and deep copy are two methods used to create copies of objects such as lists.
<p>The main difference is how they handle nested objects (lists inside another list).

<p>1. Shallow Copy:A shallow copy creates a new list, but the inner objects are not copied separately. Both the original list and copied list refer to the same nested objects.

<p>We can create a shallow copy using:copy() method,Slicing ([:]),list() function, copy.copy() function

```python
import copy

list1 = [[10, 20], [30, 40]]

list2 = copy.copy(list1)

list2[0][0] = 100

print("Original list:", list1)
print("Copied list:", list2)
```

```python
#2. Deep Copy:
#A deep copy creates a completely independent copy of the original list, including its nested objects.
import copy

list1 = [[10, 20], [30, 40]]

list2 = copy.deepcopy(list1)

list2[0][0] = 100

print("Original list:", list1)
print("Copied list:", list2)
```

```python
import copy

list1 = [[1, 2], [3, 4]]

# Shallow copy
list2 = copy.copy(list1)

# Deep copy
list3 = copy.deepcopy(list1)

list2[0][0] = 100
list3[1][0] = 200

print("Original:", list1)
print("Shallow copy:", list2)
print("Deep copy:", list3)
```
