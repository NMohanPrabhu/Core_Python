```python
#Get dimensions of first matrix
r1 = int(input("Enter rows of first matrix: "))
c1 = int(input("Enter columns of first matrix: "))

# Get dimensions of second matrix
r2 = int(input("Enter rows of second matrix: "))
c2 = int(input("Enter columns of second matrix: "))

# Check whether multiplication is possible
if c1 != r2:
    print("Matrix multiplication is not possible.")
else:
    # First matrix
    A = []
    print("Enter elements of first matrix:")

    for i in range(r1):
        row = []
        for j in range(c1):
            value = int(input(f"A[{i}][{j}]: "))
            row.append(value)
        A.append(row)

    # Second matrix
    B = []
    print("Enter elements of second matrix:")

    for i in range(r2):
        row = []
        for j in range(c2):
            value = int(input(f"B[{i}][{j}]: "))
            row.append(value)
        B.append(row)

    # Create result matrix with zeros
    C = []

    for i in range(r1):
        row = []
        for j in range(c2):
            row.append(0)
        C.append(row)

    # Matrix multiplication
    for i in range(r1):
        for j in range(c2):
            for k in range(c1):
                C[i][j] = C[i][j] + A[i][k] * B[k][j]

    # Print result
    print("Result of matrix multiplication:")

    for i in range(r1):
        for j in range(c2):
            print(C[i][j], end=" ")
        print()
```
