<font color="blue" size="6">Prime check using a function</font>

```python
# Function to check prime number
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
# Get input from user
n = int(input("Enter a number: "))
# Check the result
if is_prime(n):
    print(n, "is a prime number")
else:
    print(n, "is not a prime number")
```
