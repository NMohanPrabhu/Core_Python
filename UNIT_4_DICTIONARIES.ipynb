<p>1. What is a Dictionary?

<p>A dictionary in Python is a collection of data stored in the form of key-value pairs.

<p>    Ordered – Maintains insertion order.

<p>    Mutable – Elements can be added, modified, and deleted.

<p>    Unique keys – Duplicate keys are not allowed.

<p>    Values can be duplicated.

 <p>   Different data types can be stored.

<p>Dictionaries are created using curly brackets {}.

```python
#2. Creating Dictionaries
# Empty dictionary
d1 = {}

# Dictionary with integer values
d2 = {1: 100, 2: 200, 3: 300}

# Dictionary with string values
d3 = {"name": "Ram", "city": "Chennai"}

# Dictionary with different data types
d4 = {"name": "Sam", "age": 21, "mark": 95.5}
```

```python
#3. Operations on Dictionaries
#1. Accessing Elements
student = {"name": "Arun", "age": 20}

print(student["name"])
print(student["age"])
```

```python
#Using get():
student = {"name": "Arun", "age": 20}

print(student.get("name"))
print(student.get("city"))
```

```python
#2. Adding Elements
#We can add new key-value pairs to a dictionary.
student = {"name": "Arun", "age": 20}

student["city"] = "Coimbatore"

print(student)
```

```python
#3. Updating Elements
#We can change the value of an existing key.
student = {"name": "Arun", "age": 20}

student["age"] = 21

print(student)
```

```python
#4. Removing Elements
#Using pop():
#Removes the specified key and returns its value.
student = {"name": "Arun", "age": 20}
student.pop("age")
print(student)
```

```python
#Using popitem():
#Removes the last inserted key-value pair.
student = {"name": "Arun", "age": 20}
student.popitem()
print(student)
```

```python
#Using del:
#Deletes a particular key-value pair.
student = {"name": "Arun", "age": 20}
del student["name"]
print(student)
```

```python
#Using clear():
#Removes all elements from the dictionary.
student = {"name": "Arun", "age": 20}
student.clear()
print(student)
```

```python
#5. keys()
#Returns all keys in a dictionary.
student = {"name": "Arun", "age": 20, "city": "Chennai"}
print(student.keys())
```

```python
#6. values():Returns all values in a dictionary.
student = {"name": "Arun", "age": 20, "city": "Chennai"}
print(student.values())
```

```python
#7. items():Returns all key-value pairs as tuples.
student = {"name": "Arun", "age": 20}
print(student.items())
```

```python
#8. update():Updates an existing dictionary with new key-value pairs.
student = {"name": "Arun", "age": 20}
student.update({"age": 21, "city": "Chennai"})
print(student)
```

```python
#9. len():Returns the number of key-value pairs.
student = {"name": "Arun", "age": 20, "city": "Chennai"}
print(len(student))
```

```python
#10. copy():Creates a shallow copy of a dictionary.
student = {"name": "Arun", "age": 20}
new_student = student.copy()
print(new_student)
```

```python
#11. setdefault():Returns the value of a specified key. 
#If the key does not exist, it inserts the key with the given default value.
student = {"name": "Arun", "age": 20}

student.setdefault("city", "Coimbatore")

print(student)
```

```python
#12. fromkeys()
#Creates a new dictionary with specified keys and a common value.
keys = ("name", "age", "city")

student = dict.fromkeys(keys, 0)

print(student)
```

```python
#4. Dictionary Iteration
#We can use a for loop to access dictionary elements.
student = {"name": "Arun", "age": 20}

for x in student:
    print(x)
```

```python
#Example 2: Print values
student = {"name": "Arun", "age": 20}

for x in student.values():
    print(x)
```

```python
#Example 3: Print keys and values
student = {"name": "Arun", "age": 20}
for key, value in student.items():
    print(key, value)
```

```python
#Program 2: Find the sum of dictionary values
d = {"a": 10, "b": 20, "c": 30}

total = sum(d.values())

print("Sum =", total)
```

```python
#Program 3: Find the maximum value
d = {"a": 10, "b": 50, "c": 30}

print(max(d.values()))
```

```python
#Program 4: Check whether a key exists
d = {"name": "Arun", "age": 20}
if "name" in d:
    print("Key exists")
else:
    print("Key not found")
```

```python
#Program 5: Merge two dictionaries
d1 = {"a": 10, "b": 20}
d2 = {"c": 30, "d": 40}
d1.update(d2)
print(d1)
```
