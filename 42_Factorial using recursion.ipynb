<font color="blue" size="6"> Factorial using recursion </font>

```python
# Function to calculate factorial
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)


# Get input from user
n = int(input("Enter a number: "))

# Calculate factorial
result = factorial(n)

print("Factorial:", result)
```
