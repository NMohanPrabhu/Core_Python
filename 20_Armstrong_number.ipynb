<font color="blue" size="6">Check Armstrong number</font>

```python
n = int(input("Enter a number: "))
original = n
digits = len(str(n))
sum = 0
while n > 0:
    digit = n % 10
    sum += digit ** digits
    n //= 10
if original == sum:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```
