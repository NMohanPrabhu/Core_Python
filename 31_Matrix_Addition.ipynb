<font color="blue" size="6">MAtrix Addition </font>

```python
# Get matrix size
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

# First matrix
A = []
print("Enter elements of first matrix:")

for i in range(rows):
    row = []
    for j in range(cols):
        value = int(input(f"A[{i}][{j}]: "))
        row.append(value)
    A.append(row)

# Second matrix
B = []
print("Enter elements of second matrix:")

for i in range(rows):
    row = []
    for j in range(cols):
        value = int(input(f"B[{i}][{j}]: "))
        row.append(value)
    B.append(row)

# Addition matrix
C = []

for i in range(rows):
    row = []
    for j in range(cols):
        row.append(A[i][j] + B[i][j])
    C.append(row)

# Print result
print("Result of matrix addition:")

for i in range(rows):
    for j in range(cols):
        print(C[i][j], end=" ")
    print()

```
