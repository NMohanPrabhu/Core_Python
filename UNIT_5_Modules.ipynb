<font color="red" size="6">Modules</font>

<p>A module in Python is a file that contains Python code such as functions, variables, and classes. We can reuse this code in another Python program by importing the module.
<p>Suppose you are developing a calculator program. Instead of writing addition, subtraction, multiplication, and division code in every program, you can create a separate module and reuse it whenever needed.

#Creating a module
#Create a Python file named mymodule.py.
mymodule.py
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def div(a, b):
    return a / b 
<p> <b>Note:I have already created this module in the currect directory</b>

```python
#Importing a module

import mymodule

print(mymodule.add(10, 5))
print(mymodule.sub(10, 5))
```

<p>Explanation:
<p>•	import mymodule imports the module.
<p>•	mymodule.add(10, 5) calls the addition function.
<p>•	mymodule.sub(10, 5) calls the subtraction function.
<p>•	Similarly, multiplication and division functions are called.


```python
#Types of Modules in Python
#A. Built-in Modules:Built-in modules are modules provided by Python. We can use them directly by importing them
#Example 1: Math Module
import math

print(math.sqrt(25))
print(math.factorial(5))
print(math.pi)

```

```python
#Example 2: Random Module
import random

print(random.randint(1, 10))
```

```python
#Different Ways to Import Modules
#A. Import entire module
import math
print(math.sqrt(64))
```

```python
#C. Import with an alias
#We can give a shorter name to a module using the as keyword.
import math as m
print(m.sqrt(49))
```

```python
#D. Import all functions
#We can import all public names from a module using *.
from math import *
print(sqrt(81))
print(factorial(5))
```
