<font color="blue" size="6">Search for an element</font>

```python
# Get the number of elements
n = int(input("Enter the number of elements: "))

# Create an array
arr = []

# Get the values
for i in range(n):
    value = int(input(f"Enter element {i + 1}: "))
    arr.append(value)

# Element to search
search = int(input("Enter the element to search: "))

# Search the array
found = False

for i in range(n):
    if arr[i] == search:
        found = True
        print("Element found at position:", i + 1)
        break

if not found:
    print("Element not found")
```
