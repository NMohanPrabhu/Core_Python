<font color="red" size="6">WOrd Count</font>

```python
# Open the file in read mode
file = open("sample.txt", "r")

# Read the file content
content = file.read()

# Count the words
words = content.split()
word_count = len(words)

# Display the result
print("Total number of words:", word_count)

# Close the file
file.close()
```

Explanation

    open() opens the file in read mode.

    read() reads the entire file content.

    split() separates the content into words.

    len() counts the number of words.

    close() closes the file.

<font color="red" size="6">Python Program to Count Unique Words and Display Their Frequency</font>

```python
# Open the file in read mode
with open("sample.txt", "r") as file:
    content = file.read()

# Split the content into words
words = content.lower().split()

# Create an empty dictionary
frequency = {}

# Count the frequency of each word
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# Display the number of unique words
print("Number of unique words:", len(frequency))

# Display each word and its frequency
print("Word Frequency:")
for word, count in frequency.items():
    print(word, ":", count)
```

Explanation

    read() reads the entire file.

    lower() converts all words into lowercase.

    split() separates the text into individual words.

    A dictionary stores each unique word and its frequency.

    len(frequency) gives the number of unique words.

    items() displays each word with its frequency.
