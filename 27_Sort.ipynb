```python
# Get the number of elements
n = int(input("Enter the number of elements: "))

# Create an array
arr = []

# Get the values
for i in range(n):
    value = int(input(f"Enter element {i + 1}: "))
    arr.append(value)

# Bubble sort
for i in range(n):
    for j in range(0, n - i - 1):
        if arr[j] > arr[j + 1]:
            # Swap the elements
            temp = arr[j]
            arr[j] = arr[j + 1]
            arr[j + 1] = temp

print("Sorted array:", arr)
```

```python
#Using Built In function
# Get the number of elements
n = int(input("Enter the number of elements: "))

# Create an array
arr = []

# Get the values
for i in range(n):
    value = int(input(f"Enter element {i + 1}: "))
    arr.append(value)

# Sort the array
arr.sort()

print("Sorted array:", arr)
```
