<p><b>Types of Functions in Python</b></p>

<p><b>Python functions are mainly divided into four types:</b></p>

 <ol>
  <li>Built-in functions</li>
  <li>User-defined functions</li>
  <li>Anonymous (lambda) functions</li>
   <li>Recursive functions</li>
  </ol>   

    

```python
#User-Defined Functions:A user-defined function is a function created by the programmer using the def keyword.
def greet():
    print("Hello, Welcome to Python!")

greet()
```

Explanation:

    def – Keyword used to define a function.

    greet – Function name.

    () – Parentheses used for parameters.

    : – Indicates the beginning of the function body.

    greet() – Calls the function.

```python
#Function with Parameters:Parameters are variables passed inside the function definition.
def add(a, b):
    print("Addition:", a + b)

add(10, 20)
```

Explanation:

    a and b are parameters.

    10 and 20 are arguments.

    The function adds the two values.

```python
#Function with Return Statement:The return statement sends a result back to the function caller.
def multiply(a, b):
    return a * b

result = multiply(5, 4)

print("Result:", result)
```

```python
#Types of Function Arguments:Python supports different ways of passing arguments.
#A. Positional Arguments:Arguments are passed according to their position.
def student(name, age):
    print(name)
    print(age)

student("Arun", 21)
```

```python
#Keyword Arguments:Arguments are passed using parameter names.
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student(age=21, name="Arun")
```

```python
#Default Arguments:A default value is assigned to a parameter in the function definition.
def greet(name="Student"):
    print("Hello", name)
greet()
greet("Arun")
```

```python
#Variable-Length Arguments:These allow a function to accept multiple arguments.
#1. *args:Used to pass multiple positional arguments.
def add(*numbers):
    total = 0

    for n in numbers:
        total += n

    print("Total:", total)

add(10, 20, 30, 40)
```

```python
#2. **kwargs:Used to pass multiple keyword arguments.
def details(**data):
    for key, value in data.items():
        print(key, ":", value)

details(name="Arun", age=21, city="Coimbatore")
```

```python
#6. Lambda Function

#A lambda function is a small anonymous function without a regular function name.

#It is created using the lambda keyword.

#Syntax: lambda arguments: expression
square = lambda x: x * x

print(square(5))
```

```python
add = lambda a, b: a + b

print(add(10, 20))
```

```python
#7. Recursive Function

#A recursive function is a function that calls itself.

#Example: Factorial
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))
```

```python
#Function Scope: Local and Global Variables
#Local Variable:A variable declared inside a function is called a local variable.
def display():
    x = 10
    print(x)

display()
```

```python
#Global Variable:A variable declared outside a function is called a global variable.
x = 100

def display():
    print(x)

display()
print(x)
```

```python
#To modify a global variable inside a function, use the global keyword.
x = 10

def change():
    global x
    x = 50

change()
print(x)
```

```python
#Practical Example – Calculator Using Functions
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b

print("Addition:", add(10, 5))
print("Subtraction:", subtract(10, 5))
print("Multiplication:", multiply(10, 5))
print("Division:", divide(10, 5))
```
