<font color="Blue" size="6">Count the frequency of each character in a string</font>

```python
text = input("Enter a string: ")
frequency = {}
for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1
print("Character frequency:")
for char, count in frequency.items():
    print(char, ":", count)
```
