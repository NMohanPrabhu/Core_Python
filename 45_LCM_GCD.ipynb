<font color="blue" size="6">LCM AND GCD USING FUNCTIONS</font>

```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a


def lcm(a, b):
    return (a * b) // gcd(a, b)


# Input
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

# Output
print("GCD =", gcd(a, b))
print("LCM =", lcm(a, b))

```
