<p>Tuples and Operations on Tuples in Python
<p>1. What is a Tuple?

<p>A tuple in Python is a collection of elements that is:

<p>    Ordered – Elements maintain their insertion order.

 <p>   Immutable – Elements cannot be changed after creation.

 <p>   Allows duplicates – The same value can appear multiple times.

 <p>   Allows different data types – Integers, strings, floats, etc.

<p>Tuples are created using parentheses ().

```python
#Creating Tuples
# Empty tuple
t1 = ()

# Tuple with integers
t2 = (10, 20, 30)

# Tuple with strings
t3 = ("Apple", "Banana", "Mango")

# Tuple with different data types
t4 = (10, "Python", 3.5)

# Tuple without parentheses
t5 = 10, 20, 30

# Single-element tuple
t6 = (10,)
```

```python
#Operations on Tuples
#$. Indexing
#Accessing elements using their position.
t = (10, 20, 30, 40, 50)

print(t[0])
print(t[2])
print(t[-1])
```

```python
#Slicing
#Accessing a specific range of elements.
#Syntax:tuple[start:end:step]
t = (10, 20, 30, 40, 50)
print(t[1:4])
print(t[:3])
print(t[2:])
print(t[::-1])
```

```python
#3. Concatenation (+)
t1 = (10, 20, 30)
t2 = (40, 50, 60)

t3 = t1 + t2

print(t3)
```

```python
#4. Repetition (*):Repeating tuple elements.
t = (10, 20, 30)

print(t * 3)
```

```python
#5. Membership (in and not in):Checking whether an element exists in a tuple.
t = (10, 20, 30, 40)
print(20 in t)
print(50 not in t)
```

```python
#6. Length (len):Finding the number of elements.
t = (10, 20, 30, 40)
print(len(t))
```

```python
#7. Minimum (min):Finding the smallest element.
t = (10, 20, 5, 40)
print(min(t))
```

```python
#8. Maximum (max):Finding the largest element.
t = (10, 20, 5, 40)
print(max(t))
```

```python
#9. Sum:Finding the sum of all elements.
t = (10, 20, 30, 40)
print(sum(t))
```

```python
#10. Count():Counts how many times an element occurs.
t = (10, 20, 10, 30, 10)
print(t.count(10))
```

```python
#11. Index():Returns the index position of an element's first occurrence.
t = (10, 20, 30, 40)
print(t.index(30))
```

```python
#12. Sorting:The sorted() function sorts tuple elements and returns a list.
t = (40, 10, 30, 20)
print(sorted(t))
```

```python
#Important Point: Tuple is Immutable:Once a tuple is created, we cannot directly modify, add, or remove its elements.
t = (10, 20, 30)

t[1] = 50

print(t)
```
