<font color="blue" size="6">Print prime numbers in a range</font>

```python
start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))
for n in range(start, end + 1):
    if n > 1:
        prime = True
        for i in range(2, n):
             if n % i == 0:
                prime = False
                break
        if prime:
            print(n, end=" ")
```

The main idea
Think of it like this:

Take a number
      ↓
Is it greater than 1?
      ↓
Check divisibility from 2 to n-1
      ↓
Divisible by any number?
   ↙          ↘
 YES           NO
  ↓             ↓
Not Prime     Prime
                ↓
              Print

Key point: n % i gives the remainder. If the remainder is 0, n is exactly divisible by i, so it is not prime.
