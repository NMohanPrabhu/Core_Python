```python
#Aim: To create a list using list comprehension and display the elements.
# Create a list of numbers from 1 to 10
numbers = [i for i in range(1, 11)]

print("Original List:", numbers)

# Create a list of even numbers
even = [i for i in numbers if i % 2 == 0]

print("Even Numbers:", even)

# Create a list of squares
squares = [i * i for i in numbers]

print("Squares:", squares)
```

Explanation

    List comprehension is used to create a new list in a single line.

    for loop is used to iterate through elements.

    if condition is used to filter elements.

    i * i calculates the square of each number.

    range(1, 11) generates numbers from 1 to 10.

Result: Thus, the Python program demonstrates list comprehension to generate a list, filter even numbers, and calculate squares.

```python
#Converting words into uppercase
words = ["python", "java", "c", "c++"]

upper = [word.upper() for word in words]

print(upper)
```

```python
#List comprehension with if-else condition
numbers = [1, 2, 3, 4, 5, 6]

result = ["Even" if i % 2 == 0 else "Odd" for i in numbers]

print(result)
```

```python
#Difference Between Normal For Loop and List Comprehension
numbers = []

for i in range(1, 6):
    numbers.append(i)

print(numbers)
```

```python
#Using list comprehension:
numbers = [i for i in range(1, 6)]

print(numbers)
```

Advantages of List Comprehension

    Simple and easy to understand.

    Reduces program length.

    Improves code readability when used appropriately.

    Supports conditional statements.

    Helps perform mathematical operations.

    Creates a new list without modifying the original list.
