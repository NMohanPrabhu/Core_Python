<font color="red" size="6">Python Operators – Detailed Explanation</font>

<p><b>An operator in Python is a special symbol or keyword used to perform operations on variables and values.

```python
#Arithmetic Operators:Arithmetic operators are used to perform mathematical calculations.
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)
print("Floor Division:", a // b)
print("Power:", a ** b)
```

```python
#Comparison (Relational) Operators:Comparison operators are used to compare two values. They return either True or False.
a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

```python
#Assignment Operators:Assignment operators are used to assign values to variables and update their values.
a = 10

a += 5
print(a)

a -= 3
print(a)

a *= 2
print(a)
```

```python
#Logical Operators:Logical operators are used to combine conditional statements.
a = 10
b = 20

print(a > 5 and b > 10)
print(a > 15 or b > 10)
print(not(a > 5))
```

<b>Bitwise Operators:Bitwise operators perform operations on the binary representation of integers.
    

```python
a = 5
b = 3

print(a & b)
print(a | b)
print(a ^ b)
print(~a)
print(a << 1)
print(a >> 1)
```

```python
#Explanation:

#    5 in binary = 0101

 #   3 in binary = 0011
#    0101
#  & 0011
#  ------
#    0001
#Therefore, the answer is 1.
```

<p>7. Membership Operators

<P>Membership operators are used to check whether a value exists in a sequence, such as a string, list, tuple or set.

<P>There are two membership operators:

<P>    in – Returns True if the value exists.

 <P>   not in – Returns True if the value does not exist.

```python
fruits = ["apple", "banana", "orange"]

print("apple" in fruits)
print("mango" in fruits)
print("mango" not in fruits)
```

<p>Identity Operators

<p>Identity operators are used to check whether two variables refer to the same object in memory.

<p>There are two identity operators:

<p>    is – Returns True if both variables refer to the same object.

 <p>   is not – Returns True if they refer to different objects.

```python
a = [10, 20, 30]
b = a
c = [10, 20, 30]

print(a is b)
print(a is c)
print(a == c)
print(a is not c)
```

<p>Difference Between == and is
<p> == Comparison operator,Checks values,Two different objects can be equal</p>
<p>is,Identity operator,Checks object identity,Both variables must refer to the same object</p>

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)
print(a is b)
```
