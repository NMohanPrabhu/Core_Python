<font color="red" size="6">FILE HANDLING</font>

```python
#Reading a File: Reading means retrieving the data already stored in a file.
#Note: there must a file already existing and there must be some contents.
file = open("student.txt", "r")
data = file.read()
print(data)
file.close()
```

<p>Explanation:
<p>•	open() opens the file in reading mode.
<p>•	read() reads the entire file content.
<p>•	print() displays the content.
<p>•	close() closes the file.


```python
#Example 2: Using readline()
#Explanation: The readline() method reads only one line at a time.
file = open("student.txt", "r")

data = file.readline()

print(data)

file.close()
```

```python
#Example 3: Using readlines()

file = open("student.txt", "r")

data = file.readlines()

print(data)

file.close()
#Explanation: The readlines() method reads all lines and stores them in a list.
```

```python
#Writing into a File
#Writing means storing new data inside a file.
#Python uses the write() method to write data into a file.
file = open("student.txt", "w")

file.write("Name: Arun\n")
file.write("Age: 20\n")
file.write("Department: Computer Science")

file.close()
```

<p>Explanation:
<p>•	open("student.txt", "w") opens the file in writing mode.
<p>•	write() inserts data into the file.
<p>•	\n moves the next data to a new line.
<p>•	close() closes the file.
<p>Important If the file already contains data, using w mode deletes the old content and writes the new content.


```python
#Appending Data to a File
#Appending means adding new data at the end of an existing file without deleting the previous content.
#Python uses a mode for appending.
file = open("student.txt", "a")

file.write("\nMarks: 85")
file.write("\nGrade: A")

file.close()
```

<p>Explanation:
<p>•	a opens the file in append mode.
<p>•	write() adds new information at the end.
<p>•	Existing information remains unchanged.
<p>•	close() closes the file.


```python
#Creating a New File
#Python provides the x mode to create a new file.
file = open("newfile.txt", "x")

print("File created successfully")

file.close()
```

<p>Explanation:
<p>•	x creates a new file.
<p>•	If the file already exists, Python raises a FileExistsError.
<p>•	It does not overwrite an existing file.


```python
#Closing a File
#After performing file operations, it is important to close the file.
#Python uses the close() method.
file = open("student.txt", "r")

data = file.read()

print(data)

file.close()
```

<p>Why do we close a file?
<p>•	Releases system resources.
<p>•	Prevents unnecessary file usage.
<p>•	Helps ensure written data is properly saved.
<p>•	Avoids problems when accessing the file again.


```python
#Using the with Statement
#Python provides a simpler and safer way to work with files using the with statement.
#It automatically closes the file after the block finishes, even if an exception occurs.
#Example:
with open("student.txt", "r") as file:
    data = file.read()
    print(data)
```

```python
#Deleting a File
#Python provides the remove() function from the os module to delete a file.
import os
os.remove("student.txt")
print("File deleted successfully")
```

<p>Explanation:
<p>•	import os imports the operating system module.
<p>•	os.remove() deletes the specified file.
<p>•	print() displays a confirmation message.


```python
#File Operations – Complete Program
# 1. Creating and writing a file

file = open("student.txt", "w")

file.write("Name: Arun\n")
file.write("Age: 20\n")
file.write("Department: CSE\n")

file.close()

print("File created and data written")


# 2. Reading the file

file = open("student.txt", "r")

print("\nStudent Details:")

data = file.read()
print(data)

file.close()


# 3. Appending new data

file = open("student.txt", "a")

file.write("Marks: 90\n")

file.close()

print("New data added")


# 4. Reading updated content

file = open("student.txt", "r")

print("\nUpdated Student Details:")
print(file.read())

file.close()
```
