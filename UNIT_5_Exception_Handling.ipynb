<font color="red" size="6">Exception Handling </font>

<p>Exception handling in Python is a mechanism used to handle runtime errors so that a program does not terminate unexpectedly.
<p>Python provides four important keywords for exception handling:
 <p><b>   Python provides four important keywords for exception handling:</b></p>
<ol>
    <li>try</li>
    <li>except</li>
    <li>else</li>
    <li> finally</li>
</ol>

<p><b>1. try block:The try block contains code that might generate an error.</b></p>
<p><b>2. except block:The except block handles the error if an exception occurs.</b></p>
<p><b>3. else block:The else block executes only when no exception occurs in the try block.</b></p> 
<p><b>4. finally block:The finally block always executes, whether an exception occurs or not.</b></p> 

```python
#Example 1:
try:
    # Code that may produce an error
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2

except ZeroDivisionError:
    # Executes if division by zero occurs
    print("Error: Cannot divide by zero!")

except ValueError:
    # Executes if the user enters invalid input
    print("Error: Please enter numbers only!")

else:
    # Executes if no exception occurs
    print("Division result:", result)

finally:
    # Always executes
    print("Program execution completed.")
```

#Explanation of each block
<p>A. try block
<p>The try block contains the code that may produce an error.

<p>Here, the program accepts two numbers from the user.

<p>Then it performs division.

<p>If an error occurs, Python immediately moves to the matching except block.

<p>If no error occurs, Python moves to the else block.

<p>B. except block
<p>The except block is used to handle exceptions.

<p>Here, we have used two except blocks.

<p>First except – ZeroDivisionError

<p>    Occurs when the user enters zero as the second number.

<p>    Division by zero is not possible.

<p>    So, Python displays the error message.

<p>Second except – ValueError

<p>    Occurs when the user enters invalid input, such as abc.

<p>    Python cannot convert abc into an integer.

<p>    So, Python displays the appropriate error message.
    
<p>C. else block
<p>    The else block executes only when no exception occurs in the try block.

<p>    If the user enters valid numbers and the division is successful, the result is displayed.

<p>    If an exception occurs, the else block is skipped.
<p>D. finally block
<p>    The finally block always executes.

<p>    It executes whether an exception occurs or not.

<p>    It is commonly used for cleanup operations, such as closing files or database connections.

Important points to remember

    A try block must be followed by at least one except or finally block.

    We can use multiple except blocks.

    The else block is optional.

    The finally block is optional.

    The else block executes only when the try block completes without an exception.

    The finally block executes whether an exception occurs or not.

```python
#Raising an Exception (raise):The raise keyword is used to generate an exception manually.
age = 15

if age < 18:
    raise ValueError("Age must be 18 or above")
else:
    print("Eligible")
```

Advantages of Exception Handling

    Prevents unexpected program termination.

    Makes programs more reliable.

    Helps identify and debug errors.

    Separates error-handling code from normal code.

    Ensures cleanup operations using finally.
