<font color="red" size="6">Copy a file </font>

```python
try:
    # Open the source file and read its contents
    with open("sample.txt", "r") as file:
        content = file.read()

    # Copy the contents into another file
    with open("copy.txt", "w") as file:
        file.write(content)

    print("File copied successfully!")

    # Display the copied file contents
    with open("copy.txt", "r") as file:
        print("\nContents of the copied file:")
        print(file.read())

except FileNotFoundError:
    print("Error: Source file not found.")

except Exception as e:
    print("An error occurred:", e)
```

Explanation

    open("sample.txt", "r") – Opens the source file in read mode.

    read() – Reads the contents of the file.

    open("copy.txt", "w") – Creates a new file in write mode.

    write() – Copies the contents into the new file.

    FileNotFoundError – Handles the error if the source file does not exist.

    Finally, the copied file is opened and its contents are displayed.

Result: Thus, the Python program copies the contents of one file into another file and displays the copied contents.
