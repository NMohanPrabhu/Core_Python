```python
# Get the number of elements
n = int(input("Enter the number of elements: "))

# Create an array
arr = []

# Get the values
for i in range(n):
    value = int(input(f"Enter element {i + 1}: "))
    arr.append(value)

# Reverse the array
arr.reverse()

print("Reversed array:", arr)
```

```python
arr = arr[::-1]
```

```python
print("Reversed array:", arr)
```

```python
#WITHOUT USING FUNCTIONS
# Get the number of elements
n = int(input("Enter the number of elements: "))

# Create an array
arr = []

# Get the values
for i in range(n):
    value = int(input(f"Enter element {i + 1}: "))
    arr.append(value)

# Reverse the array using a loop
for i in range(n // 2):
    temp = arr[i]
    arr[i] = arr[n - 1 - i]
    arr[n - 1 - i] = temp

print("Reversed array:", arr)
```

```python
#We use n // 2 because each swap reverses two elements, so we only need to go through half of the array.
```
