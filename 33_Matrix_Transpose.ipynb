```python
# Get matrix size
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

# Create matrix
A = []

print("Enter matrix elements:")

for i in range(rows):
    row = []
    for j in range(cols):
        value = int(input(f"A[{i}][{j}]: "))
        row.append(value)
    A.append(row)

# Create transpose matrix
T = []

for j in range(cols):
    row = []
    for i in range(rows):
        row.append(A[i][j])
    T.append(row)

# Print original matrix
print("Original matrix:")

for i in range(rows):
    for j in range(cols):
        print(A[i][j], end=" ")
    print()

# Print transpose
print("Transpose matrix:")

for i in range(cols):
    for j in range(rows):
        print(T[i][j], end=" ")
    print()
```
