<font color="blue" size="6"> READ WRITE AND MODIFY AN ELEMENT IN ARRAY</font>

```python
# Create an array from user input
arr = input("Enter elements separated by spaces: ").split()
print("Array:", arr)

# Append an element
new_item = input("Enter an element to append: ")
arr.append(new_item)
print("After append:", arr)

# Rename an element
old_item = input("Enter the element you want to rename: ")
new_name = input("Enter the new name: ")

if old_item in arr:
    index = arr.index(old_item)
    arr[index] = new_name
    print("After rename:", arr)
else:
    print("Element not found.")

# Delete an element
delete_item = input("Enter the element you want to delete: ")

if delete_item in arr:
    arr.remove(delete_item)
    print("After delete:", arr)
else:
    print("Element not found.")

print("Final array:", arr)
```

```python
#If you want the user to first enter the number of elements, and then enter each value,
# Get the number of elements
n = int(input("Enter the number of elements: "))

# Create an empty array
arr = []

# Get the values
for i in range(n):
    value = input(f"Enter element {i + 1}: ")
    arr.append(value)

# Print the array
print("Array:", arr)
```
