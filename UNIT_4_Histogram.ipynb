```python
#A histogram represents the frequency of each value using bars.
numbers = [10, 20, 20, 30, 30, 30, 40, 40, 50]

for i in numbers:
    if numbers.count(i) > 0:
        print(i, ":", "*" * numbers.count(i))
        numbers = [x for x in numbers if x != i]
```

```python
#2. Histogram program using a dictionary
numbers = [10, 20, 20, 30, 30, 30, 40, 40, 50]

frequency = {}

for i in numbers:
    if i in frequency:
        frequency[i] += 1
    else:
        frequency[i] = 1

for i in frequency:
    print(i, ":", "*" * frequency[i])
```

Explanation

    frequency = {} – Creates an empty dictionary.

    for i in numbers – Reads each element from the list.

    if i in frequency – Checks whether the element already exists.

    frequency[i] += 1 – Increases its frequency by one.

    print(i, ":", "*" * frequency[i]) – Displays the histogram using stars.
