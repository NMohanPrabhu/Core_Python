<font color="red" size="6">Voter's Age and Marks Validation Using Exception Handling</font>

```python
try:
    age = int(input("Enter your age: "))

    if age < 0:
        raise ValueError("Age cannot be negative")

    if age >= 18:
        print("You are eligible to vote.")
    else:
        print("You are not eligible to vote.")

except ValueError as e:
    print("Invalid input:", e)

except Exception:
    print("An unexpected error occurred.")
```

Explanation

    try: Contains the code that may generate an exception.

    int(): Converts the entered age into an integer.

    raise ValueError: Raises an exception if the age is negative.

    except ValueError: Handles invalid input.

    except Exception: Handles any unexpected errors.

    If age is 18 or above, the person is eligible to vote.

<font color="red" size="6">Python Program for Marks Range Validation Using Exception Handling</font>

```python
try:
    marks = int(input("Enter your marks: "))

    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100")

    print("Valid marks:", marks)

    if marks >= 50:
        print("Result: Pass")
    else:
        print("Result: Fail")

except ValueError as e:
    print("Invalid input:", e)

except Exception:
    print("An unexpected error occurred.")
```

Explanation

    try – Executes the statements that may cause an exception.

    int() – Converts the input into an integer.

    raise ValueError – Raises an exception if marks are below 0 or above 100.

    except ValueError – Handles invalid marks and non-numeric input.

    If marks are 50 or above, the student passes; otherwise, the student fails.
