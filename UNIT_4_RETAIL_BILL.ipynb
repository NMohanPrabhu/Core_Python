#Retail Bill Preparation Using Python (List and Tuple)
# Retail Bill Preparation

# Item details using tuple
items = (
    ("Pen", 10, 5),
    ("Book", 50, 2),
    ("Pencil", 5, 10),
    ("Bag", 500, 1)
)

total = 0

print("       RETAIL BILL")
print("------------------------------")
print("Item      Price  Qty  Amount")
print("------------------------------")

for item in items:
    name = item[0]
    price = item[1]
    quantity = item[2]

    amount = price * quantity
    total = total + amount

    print(name, price, quantity, amount)

print("------------------------------")
print("Total Amount:", total)

discount = total * 0.10
net_amount = total - discount

print("Discount (10%):", discount)
print("Net Amount:", net_amount)
print("------------------------------")
print("Thank You! Visit Again")

items – Stores item name, price and quantity using tuples.

total = 0 – Initializes the total amount.

for item in items – Reads each item one by one.

amount = price * quantity – Calculates the price of each item.

total = total + amount – Calculates the total bill.

discount = total * 0.10 – Calculates a 10% discount.

net_amount = total - discount – Calculates the final amount.
